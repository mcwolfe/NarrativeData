```python
import json
import rispy
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from collections import Counter
import re
from sklearn.feature_extraction.text import TfidfVectorizer

# Load the pre-2021 JSON file from VOSViewer
pre_2021_json_file = "../../textlab/texts/Others/wil_texter_2000_2020.json"  # Update this path
with open(pre_2021_json_file, 'r') as f:
    pre_2021_network_data = json.load(f)

# Build the NetworkX graph from VOSViewer JSON data
pre_2021_graph = nx.Graph()

# Add nodes with attributes
for node in pre_2021_network_data.get('nodes', []):
    pre_2021_graph.add_node(
        node['id'],
        label=node.get('label', ''),
        cluster=node.get('cluster', 0),
        x=node.get('x', 0),
        y=node.get('y', 0),
        weight_Links=node.get('weight_Links', 0),
        weight_Total_link_strength=node.get('weight_Total_link_strength', 0),
        weight_Documents=node.get('weight_Documents', 0)
    )

# Add edges
for edge in pre_2021_network_data.get('edges', []):
    pre_2021_graph.add_edge(
        edge['source'],
        edge['target'],
        weight=edge.get('weight', 1)
    )

# Get the largest connected component
pre_2021_components = list(nx.connected_components(pre_2021_graph))
pre_2021_components.sort(key=len, reverse=True)
pre_2021_largest = pre_2021_graph.subgraph(pre_2021_components[0]).copy()

print(f"Pre-2021 Network: {pre_2021_largest.number_of_nodes()} nodes, {pre_2021_largest.number_of_edges()} edges")

# Load the pre-2021 RIS file
pre_2021_ris_file = "../../textlab/texts/Others/pre_2021_data.ris"  # Update this path
with open(pre_2021_ris_file, 'r', encoding='utf-8') as f:
    pre_2021_entries = rispy.load(f)

# Convert to DataFrame
pre_2021_df = pd.DataFrame(pre_2021_entries)
print(f"Pre-2021 Publications: {len(pre_2021_df)}")
```
